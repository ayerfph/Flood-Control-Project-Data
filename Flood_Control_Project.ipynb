### Correct Datatypes In Datetime

```python
# Convert to datetime (from ms -> datetime) and format as YYYY-MM-DD
df['CompletionDate'] = pd.to_datetime(df['CompletionDateOriginal'], unit="ms", errors='coerce')

# Drop the old column
df = df.drop(columns=['CompletionDateOriginal'])

# Convert StartDate from string MM/DD/YYYY -> datetime
df['StartDate'] = df['StartDate'].astype(str).str.strip()
df['StartDate'] = pd.to_datetime(df['StartDate'], format='%m/%d/%Y', errors='coerce')

# Convert other date columns
df['CompletionDateActual'] = pd.to_datetime(df['CompletionDateActual'], errors="coerce")

# Ensure uniform datetime type (no premature string formatting)
# If you need to DISPLAY as YYYY-MM-DD, format when printing/exporting:
df['StartDate'] = df['StartDate'].dt.strftime('%Y-%m-%d')
df['CompletionDate'] = df['CompletionDate'].dt.strftime('%Y-%m-%d')
df['CompletionDateActual'] = df['CompletionDateActual'].dt.strftime('%Y-%m-%d')

# Show output
print(df.head(5))
```

### Getting The Fastest and Longest Construction Period

```python
# Convert CompletionDateActual into datetime
df['CompletionDateActual'] = pd.to_datetime(df['CompletionDateActual'])
df['StartDate'] = pd.to_datetime(df['StartDate'])

df['TotalDays'] = (df['CompletionDateActual'] - df['StartDate']).dt.days

#Get min and max
min_period = df['TotalDays'].min()
max_period = df['TotalDays'].max()

print("\nMinimum Construction Period (days):", min_period)
print("Maximum Construction Period (days):", max_period)
```

### Getting The Construction Period

```python
# Function to compute months + days difference
def months_days_diff(start, end):
    if pd.isna(start) or pd.isna(end):
        return None
    rd = relativedelta(end, start)   # end - start
    return f"{abs(rd.years)*12 + rd.months} months {abs(rd.days)} days"

# Apply row by row
df['ConstructionPeriod'] = df.apply(
    lambda row: months_days_diff(row['StartDate'], row['CompletionDateActual']),
    axis=1
)


df['ConstructionPeriod']
```
